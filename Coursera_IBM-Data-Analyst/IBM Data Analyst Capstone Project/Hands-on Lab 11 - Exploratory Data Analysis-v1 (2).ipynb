#### Step 3: Handling Missing Data


Identify and manage missing values in critical columns such as `Employment`, `JobSat`, and `RemoteWork`. Implement a strategy to fill or drop these values, depending on the significance of the missing data.


```python
## Write your code here

import pandas as pd

# Step 1: Check missing values in the critical columns
critical_cols = ['Employment', 'JobSat', 'RemoteWork']
missing_summary = df[critical_cols].isnull().sum()
print("Missing values before handling:\n", missing_summary)

# Step 2: Strategy for handling missing values
for col in critical_cols:
    missing_percent = df[col].isnull().mean() * 100
    
    if missing_percent < 5:  
        # If missing < 5%, drop the rows
        df.dropna(subset=[col], inplace=True)
        print(f"Dropped rows with missing values in {col}")
    
    else:
        # If missing >= 5%, fill with mode (most frequent value)
        mode_value = df[col].mode()[0]
        df[col].fillna(mode_value, inplace=True)
        print(f"Filled missing values in {col} with mode: {mode_value}")

# Step 3: Verify missing values after handling
print("\nMissing values after handling:\n", df[critical_cols].isnull().sum())

```

#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
df['JobSat'].mean()
```

```python
## Write your code here

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Step 1: Ensure YearsCodePro is numeric
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')

# Step 2: Ensure JobSat is numeric (should already be 0-10)
df['JobSat'] = pd.to_numeric(df['JobSat'], errors='coerce')

# Step 3: Define experience ranges
bins = [0, 5, 10, 20, 100]  # 100 as an upper bound
labels = ["0-5", "5-10", "10-20", "20+"]
df['ExperienceRange'] = pd.cut(df['YearsCodePro'], bins=bins, labels=labels, right=False)

# Step 4: Calculate median JobSat for each experience range
experience_jobSat = df.groupby('ExperienceRange')['JobSat'].mean().reset_index()

print("Median Job Satisfaction by Experience Range:")
print(experience_jobSat)

# Step 5: Visualization
plt.figure(figsize=(8, 5))
sns.barplot(data=experience_jobSat, x='ExperienceRange', y='JobSat', palette="magma")
plt.title("Median Job Satisfaction by Experience Range", fontsize=14)
plt.xlabel("Years of Professional Coding Experience")
plt.ylabel("Median Job Satisfaction (0-10)")
plt.ylim(0, 10)
plt.show()
```

#### Step 5: Visualize Job Satisfaction


Use a count plot to show the distribution of `JobSat` values. This provides insights into the overall satisfaction levels of respondents.


```python
## Write your code here
median_jobsat_by_range=df.groupby('ExperienceRange')['JobSat'].median().reset_index()

plt.figure(figsize=(10, 6))
median_jobsat_by_range.plot(kind='bar')
plt.xlabel('Professional Experience Range')
plt.ylabel('Median Job Satisfaction')
plt.title('Median Job Satisfaction by Professional Experience Range')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()


plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='JobSat', order=df['JobSat'].value_counts().index, palette='magma')
plt.xlabel('Job Satisfaction Score')
plt.ylabel('Count')
plt.title('Distribution of Job Satisfaction Scores')
plt.tight_layout()
plt.show()
```

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
## Write your code here

# Step 1: Ensure LanguageHaveWorkedWith is string and split multiple languages
df['LanguageHaveWorkedWith'] = df['LanguageHaveWorkedWith'].astype(str)

# Example: Filter for a specific region or country (e.g., "Canada")
region_df = df[df['Country'] == "Canada"]

# Step 2: Split languages into a list
languages_series = region_df['LanguageHaveWorkedWith'].str.split(';').explode()

# Step 3: Count frequency of each language
language_counts = languages_series.value_counts().reset_index()
language_counts.columns = ['Language', 'Count']

# Step 4: Get top N languages
top_n = 10
top_languages = language_counts.head(top_n)

# Step 5: Visualize with a bar plot
plt.figure(figsize=(10,6))
sns.barplot(data=top_languages, x='Count', y='Language', palette='magma')
plt.title(f"Top {top_n} Programming Languages in Canada", fontsize=14)
plt.xlabel("Number of Developers")
plt.ylabel("Programming Language")
plt.show()

```

```python


# Choose a few countries for comparison
df['Country'] = df['Country'].str.replace(
    "United Kingdom of Great Britain and Northern I.*", "United Kingdom", regex=True
)

countries = ["Canada", "United States of America", "India", "Germany","United Kingdom"]
heatmap_data = pd.DataFrame()

for country in countries:
    temp = df[df['Country'] == country]['LanguageHaveWorkedWith'].str.split(';').explode()
    counts = temp.value_counts()
    heatmap_data[country] = counts

# Fill missing languages with 0
heatmap_data = heatmap_data.fillna(0)

# Plot heatmap
plt.figure(figsize=(12,8))
sns.heatmap(heatmap_data, cmap='YlOrBr', linewidths=.5, annot=False, fmt='g')
plt.title("Programming Language Popularity by Region", fontsize=14)
plt.xlabel("Country")
plt.ylabel("Programming Language")
plt.show()

```

#### Step 8: Correlation Between Experience and Satisfaction


Examine how years of experience (`YearsCodePro`) correlate with job satisfaction (`JobSatPoints_1`). Use a scatter plot to visualize this relationship.


```python
## Write your code here

df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')
df['JobSatPoints_1'] = pd.to_numeric(df['JobSatPoints_1'], errors='coerce')

# Fill missing values
df['YearsCodePro'].fillna(df['YearsCodePro'].median(), inplace=True)
df['JobSatPoints_1'].fillna(df['JobSatPoints_1'].median(), inplace=True)

```

```python
df['YearsCodePro'] = df['YearsCodePro'].interpolate()
df['JobSatPoints_1'] = df['JobSatPoints_1'].interpolate()
```

```python
df['YearsCodePro'].fillna(0, inplace=True)
df['JobSatPoints_1'].fillna(df['JobSatPoints_1'].mode()[0], inplace=True)  # most common satisfaction
```

```python
correlation = df['YearsCodePro'].corr(df['JobSat'])
print(f"Correlation between YearsCodePro and JobSat is : {correlation:.2f}")

# Create a scatter plot to visualize the relationship
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='YearsCodePro', y='JobSat', alpha=0.5)
plt.xlabel('Years of Professional Coding Experience')
plt.ylabel('Job Satisfaction')
plt.title('Job Satisfaction by Experience Range')
plt.show()
```

#### Step 9: Educational Background and Employment Type


Explore how educational background (`EdLevel`) relates to employment type (`Employment`). Use cross-tabulation and visualizations to understand if higher education correlates with specific employment types.


```python
## Write your code here


# Step 1: Cross-tabulation between EdLevel and Employment
cross_tab = pd.crosstab(df['EdLevel'], df['Employment'], normalize='index') * 100

# Display as DataFrame
cross_tab_df = cross_tab.round(1)
print("Employment type distribution by education level (%):")
print(cross_tab_df)


```

```python
# Step 2: Heatmap visualization
plt.figure(figsize=(12,7))
sns.heatmap(cross_tab_df, annot=True, fmt=".1f", cmap="YlGnBu", cbar_kws={'label': 'Percentage'})
plt.title("Employment Type Distribution by Education Level", fontsize=14)
plt.xlabel("Employment Type")
plt.ylabel("Education Level")
plt.xticks(rotation=30, ha='right')
plt.show()
```

```python
# Step 3: Bar plot for easier comparison (optional)
plt.figure(figsize=(12,6))
cross_tab_df.plot(kind='bar', stacked=True, figsize=(12,6), colormap="tab20")
plt.title("Employment Type Distribution by Education Level", fontsize=14)
plt.xlabel("Education Level")
plt.ylabel("Percentage")
plt.legend(title="Employment Type", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```
