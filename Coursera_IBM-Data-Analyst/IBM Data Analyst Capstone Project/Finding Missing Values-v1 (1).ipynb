##### Task 5: Identify the most frequent (majority) value in a specific column (e.g., `Employment`).


```python
## Write your code here

df["Employment"].mode()[0]
```

##### Task 6: Impute missing values in the `Employment` column with the most frequent value.



```python
## Write your code here

# Find the most frequent value (mode)
majority_value = df["Employment"].mode()[0]

# Impute missing values with that value
df["Employment"].fillna(majority_value, inplace=True)

print(f"Missing values in 'Employment' after imputation: {df['Employment'].isnull().sum()}")
```

### 5. Visualizing Imputed Data
##### Task 7: Visualize the distribution of a column after imputation (e.g., `Employment`).


```python
## Write your code here


# Get top 10 most frequent categories
top_employment = df["Employment"].value_counts().head(10)

plt.figure(figsize=(10,6))
sns.barplot(x=top_employment.values, y=top_employment.index, palette="viridis")

plt.title("Top 10 Employment Categories After Imputation", fontsize=16)
plt.xlabel("Count")
plt.ylabel("Employment Type")
plt.show()
```
