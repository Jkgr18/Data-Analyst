### Task 1: Exploring Data Distributions Using Bubble Plots


#### 1. Bubble Plot for Age vs. Frequency of Participation


- Visualize the relationship between respondents’ age and their participation frequency (`SOPartFreq`) using a bubble plot.

- Use the size of the bubbles to represent their job satisfaction (`JobSat`).


```python
##Write your code here

df_d = df[['Age', 'SOPartFreq', 'JobSat']].dropna()

display(df_d.head())

#Group the data by Age and SOPartFreq, and calculate the average JobSat for each group.
#Reasoning: Group the data by Age and SOPartFreq and calculate the mean JobSat, then reset the index.
grouped_data = df_d.groupby(['Age', 'SOPartFreq'])['JobSat'].mean().reset_index()
display(grouped_data.head())


#Generate a bubble plot using Age on the x-axis, SOPartFreq on the y-axis, and JobSat for bubble size.

#Reasoning: Generate a bubble plot using the grouped data, mapping Age to the x-axis, SOPartFreq to the y-axis, and JobSat to the bubble size

plt.figure(figsize=(10, 6))
# Scale the bubble size by a larger factor and add a constant to ensure visibility of smaller values
plt.scatter(grouped_data['Age'], grouped_data['SOPartFreq'], s=(grouped_data['JobSat'] - grouped_data['JobSat'].min() + 1) * 200)
plt.xlabel('Age')
plt.ylabel('Stack Overflow Participation Frequency')
plt.title('Relationship between Age, Stack Overflow Participation Frequency, and Job Satisfaction')
plt.xticks(rotation=45, ha='right')
plt.grid(True)
plt.tight_layout()
plt.show()
```

#### 2. Bubble Plot for Compensation vs. Job Satisfaction


-Visualize the relationship between yearly compensation (`ConvertedCompYearly`) and job satisfaction (`JobSat`).

- Use the size of the bubbles to represent respondents’ age.


```python
import pandas as pd
import matplotlib.pyplot as plt

# --- 1. Select relevant columns and drop missing ---
df = pd.read_csv('survey_data.csv', usecols=['ConvertedCompYearly', 'JobSat', 'Age'])
df_compensation_satisfaction = df[['ConvertedCompYearly', 'JobSat', 'Age']].dropna()
display(df_compensation_satisfaction.head())



# --- 2. Group by Compensation and JobSat and get mode of Age for bubble size ---
grouped_compensation_satisfaction = (
    df_compensation_satisfaction
    .groupby(['ConvertedCompYearly', 'JobSat'])['Age']
    .agg(lambda x: x.mode()[0] if not x.mode().empty else None)
    .reset_index()
)
display(grouped_compensation_satisfaction.head())

# --- 3. Map age categories to bubble sizes ---
age_sizes = {
    'Under 18 years old': 50,
    '18-24 years old': 150,
    '25-34 years old': 250,
    '35-44 years old': 350,
    '45-54 years old': 450,
    '55-64 years old': 550,
    '65 years or older': 650,
    'Prefer not to say': 100
}

# Make sure Age values exist in age_sizes
grouped_compensation_satisfaction = grouped_compensation_satisfaction[
    grouped_compensation_satisfaction['Age'].isin(age_sizes.keys())
]

# --- 4. Plot bubble plot ---
plt.figure(figsize=(14, 10))

plt.scatter(
    grouped_compensation_satisfaction['ConvertedCompYearly'],
    grouped_compensation_satisfaction['JobSat'],
    s=grouped_compensation_satisfaction['Age'].map(age_sizes),
    alpha=0.6,
    color='purple',
    edgecolors='w'
)

plt.xscale('log')  # Log scale for compensation
plt.xlabel('Yearly Compensation (Log Scale)', fontsize=14)
plt.ylabel('Job Satisfaction', fontsize=14)
plt.title('Relationship between Compensation, Job Satisfaction, and Age', fontsize=16)
plt.grid(True, alpha=0.3)

# --- 5. Legend for age categories ---
handles = [
    plt.Line2D([], [], marker="o", linestyle="", markersize=size/20, label=age)
    for age, size in age_sizes.items()
]
plt.legend(handles=handles, title="Age")

plt.tight_layout()
plt.show()
```

#### 1. Bubble Plot of Technology Preferences by Age

- Visualize the popularity of programming languages respondents have worked with (`LanguageHaveWorkedWith`) across age groups.

- Use bubble size to represent the frequency of each language.



### Task 2: Analyzing Relationships Using Bubble Plots


```python
##Write your code here

# --- 1. Select relevant columns and drop missing ---
df = pd.read_csv('survey_data.csv')
df_languages = df[['LanguageHaveWorkedWith', 'Age']].dropna()

# --- 2. Explode multiple languages ---
df_languages['Language_list'] = df_languages['LanguageHaveWorkedWith'].str.split(';')
df_exploded = df_languages.explode('Language_list')
df_exploded['Language_list'] = df_exploded['Language_list'].str.strip()

# --- 3. Group by Language and Age to get frequency ---
language_age_freq = (
    df_exploded
    .groupby(['Language_list', 'Age'])
    .size()
    .reset_index(name='Frequency')
)

# Optional: Keep top N languages for readability
top_languages = df_exploded['Language_list'].value_counts().head(10).index.tolist()
language_age_freq = language_age_freq[language_age_freq['Language_list'].isin(top_languages)]

# --- 4. Map age categories to numeric positions (for Y-axis) ---
age_order = [
    'Under 18 years old', '18-24 years old', '25-34 years old', 
    '35-44 years old', '45-54 years old', '55-64 years old', 
    '65 years or older', 'Prefer not to say'
]
language_age_freq['Age_num'] = language_age_freq['Age'].apply(lambda x: age_order.index(x))

# --- 5. Map bubble size ---
# You can scale frequency to make bubbles visible
max_freq = language_age_freq['Frequency'].max()
language_age_freq['Bubble_size'] = language_age_freq['Frequency'] / max_freq * 1000  # scale factor

# --- 6. Create Bubble Plot ---
plt.figure(figsize=(14,8))

for _, row in language_age_freq.iterrows():
    plt.scatter(
        row['Language_list'], 
        row['Age_num'], 
        s=row['Bubble_size'], 
        alpha=0.6, 
        color='purple', 
        edgecolors='w'
    )

plt.yticks(range(len(age_order)), age_order)
plt.xlabel('Programming Language', fontsize=14)
plt.ylabel('Age Group', fontsize=14)
plt.title('Popularity of Programming Languages by Age (Bubble Plot)', fontsize=16)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

#### 2. Bubble Plot for Preferred Databases vs. Job Satisfaction

- Explore the relationship between preferred databases (`DatabaseWantToWorkWith`) and job satisfaction.

- Use bubble size to indicate the number of respondents for each database.


```python

# --- 1. Select relevant columns and drop missing ---
df_db = df[['DatabaseWantToWorkWith', 'JobSat']].dropna()

# --- 2. Explode multiple databases if respondents listed more than one ---
df_db['Database_list'] = df_db['DatabaseWantToWorkWith'].str.split(';')
df_exploded = df_db.explode('Database_list')
df_exploded['Database_list'] = df_exploded['Database_list'].str.strip()

# --- 3. Group by Database and JobSat to get frequency ---
db_jobsat_freq = (
    df_exploded
    .groupby(['Database_list', 'JobSat'])
    .size()
    .reset_index(name='Frequency')
)

# Optional: Keep top 10 databases for readability
top_databases = df_exploded['Database_list'].value_counts().head(10).index.tolist()
db_jobsat_freq = db_jobsat_freq[db_jobsat_freq['Database_list'].isin(top_databases)]

# --- 4. Scale bubble size based on frequency ---
max_freq = db_jobsat_freq['Frequency'].max()
db_jobsat_freq['Bubble_size'] = db_jobsat_freq['Frequency'] / max_freq * 1000  # adjust scale factor if needed

# --- 5. Plot Bubble Plot ---
plt.figure(figsize=(14,8))

for _, row in db_jobsat_freq.iterrows():
    plt.scatter(
        row['Database_list'], 
        row['JobSat'], 
        s=row['Bubble_size'], 
        alpha=0.6, 
        color='teal', 
        edgecolors='w'
    )

plt.xticks(rotation=45, ha='right')
plt.xlabel('Preferred Database', fontsize=14)
plt.ylabel('Job Satisfaction', fontsize=14)
plt.title('Preferred Databases vs Job Satisfaction (Bubble Plot)', fontsize=16)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

### Task 3: Comparing Data Using Bubble Plots


#### 1. Bubble Plot for Compensation Across Developer Roles

- Visualize compensation (`ConvertedCompYearly`) across different developer roles (`DevType`).

- Use bubble size to represent job satisfaction.


```python
##Write your code here


# --- 1. Select relevant columns and drop missing ---
df_roles = df[['DevType', 'ConvertedCompYearly', 'JobSat']].dropna()

# --- 2. Explode multiple developer roles ---
df_roles['DevType_list'] = df_roles['DevType'].str.split(';')
df_exploded = df_roles.explode('DevType_list')
df_exploded['DevType_list'] = df_exploded['DevType_list'].str.strip()

# --- 3. Optional: Keep top N roles for clarity ---
top_roles = df_exploded['DevType_list'].value_counts().head(10).index.tolist()
df_plot = df_exploded[df_exploded['DevType_list'].isin(top_roles)].copy()

# --- 4. Scale bubble size based on JobSat ---
# Ensure JobSat is numeric
df_plot['JobSat'] = pd.to_numeric(df_plot['JobSat'], errors='coerce')
df_plot = df_plot.dropna(subset=['JobSat'])

max_jobsat = df_plot['JobSat'].max()
df_plot['Bubble_size'] = df_plot['JobSat'] / max_jobsat * 1000  # scale factor for visibility

# --- 5. Plot Bubble Plot ---
plt.figure(figsize=(14,8))

for _, row in df_plot.iterrows():
    plt.scatter(
        row['DevType_list'], 
        row['ConvertedCompYearly'], 
        s=row['Bubble_size'], 
        alpha=0.6, 
        color='orange', 
        edgecolors='w'
    )

plt.xticks(rotation=45, ha='right', fontsize=12)
plt.xlabel('Developer Role', fontsize=14)
plt.ylabel('Yearly Compensation (USD)', fontsize=14)
plt.title('Compensation Across Developer Roles (Bubble Size = Job Satisfaction)', fontsize=16)
plt.grid(axis='y', alpha=0.3)

# Optional: log scale for compensation to handle outliers
plt.yscale('log')

plt.tight_layout()
plt.show()
```

#### 2. Bubble Plot for Collaboration Tools by Age

- Visualize the relationship between the collaboration tools used (`NEWCollabToolsHaveWorkedWith`) and age groups.

- Use bubble size to represent the frequency of tool usage.


```python
##Write your code here

import pandas as pd
import matplotlib.pyplot as plt

# --- 1. Select relevant columns and drop missing ---
df_tools = df[['NEWCollabToolsHaveWorkedWith', 'Age']].dropna()

# --- 2. Split multiple tools into rows ---
df_tools['Tool_list'] = df_tools['NEWCollabToolsHaveWorkedWith'].str.split(';')
df_exploded = df_tools.explode('Tool_list')
df_exploded['Tool_list'] = df_exploded['Tool_list'].str.strip()

# --- 3. Group by Tool and Age ---
df_grouped = df_exploded.groupby(['Tool_list', 'Age']).size().reset_index(name='Count')

# --- 4. Scale bubble size ---
df_grouped['Bubble_size'] = df_grouped['Count'] / df_grouped['Count'].max() * 1000

# --- 5. Plot Bubble Plot ---
plt.figure(figsize=(16, 8))

scatter = plt.scatter(
    df_grouped['Tool_list'],
    df_grouped['Age'],
    s=df_grouped['Bubble_size'],
    alpha=0.6,
    color="teal",
    edgecolors="w"
)

plt.xticks(rotation=45, ha='right', fontsize=12)
plt.xlabel("Collaboration Tools", fontsize=14)
plt.ylabel("Age Groups", fontsize=14)
plt.title("Bubble Plot: Collaboration Tools by Age (Bubble Size = Frequency)", fontsize=16)
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
